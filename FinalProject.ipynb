```python
import streamlit as st
import pandas as pd
import plotly.express as px
import networkx as nx
import matplotlib.pyplot as plt
import os

# Title
st.title("Twitter Political Discourse Analysis")

# Load sentiment data from mounted Google Drive path
@st.cache_data
def load_data():
    path = "/content/drive/MyDrive/twitter_data/processed_sentiment.csv"
    if not os.path.exists(path):
        st.error("Sentiment data not found in Drive at: twitter_data/processed_sentiment.csv")
        return pd.DataFrame(columns=["clean_text", "sentiment", "date"])
    df = pd.read_csv(path)
    df["date"] = pd.to_datetime(df["date"], errors='coerce')
    df = df.dropna(subset=["date", "sentiment"])
    return df

df = load_data()

if df.empty:
    st.stop()

# Sidebar filters
st.sidebar.header("Filters")
selected_sentiment = st.sidebar.multiselect(
    "Select sentiment",
    options=df["sentiment"].unique(),
    default=list(df["sentiment"].unique())
)

# Filter data
filtered_df = df[df["sentiment"].isin(selected_sentiment)]

# Sentiment by hour
st.subheader("Sentiment Trend by Hour")
filtered_df["hour"] = filtered_df["date"].dt.hour
hour_fig = px.histogram(
    filtered_df,
    x="hour",
    color="sentiment",
    barmode="group",
    title="Sentiment by Hour of Day"
)
st.plotly_chart(hour_fig)

# Sentiment distribution
st.subheader("Overall Sentiment Distribution")
dist_fig = px.histogram(
    filtered_df,
    x="sentiment",
    color="sentiment",
    title="Sentiment Distribution"
)
st.plotly_chart(dist_fig)

# Load influencer edges from Drive
@st.cache_data
def load_influencers():
    path = "/content/drive/MyDrive/twitter_data/user_edges.csv"
    if not os.path.exists(path):
        return pd.DataFrame(columns=["source", "target"])
    return pd.read_csv(path)

gdf = load_influencers()

st.subheader("Top Influencers")
if not gdf.empty:
    G = nx.DiGraph()
    G.add_edges_from(zip(gdf["source"], gdf["target"]))
    centrality = nx.degree_centrality(G)
    top_influencers = sorted(centrality.items(), key=lambda x: x[1], reverse=True)[:5]

    for user, score in top_influencers:
        st.write(f"{user}: {score:.4f}")

    st.subheader("Influencer Network")
    fig, ax = plt.subplots()
    pos = nx.spring_layout(G)
    nx.draw(G, pos, with_labels=True, node_color='skyblue', edge_color='gray',
            node_size=1500, font_size=10, ax=ax)
    st.pyplot(fig)
else:
    st.info("No influencer data available. Make sure 'user_edges.csv' is in Drive under 'twitter_data'.")

st.caption("Powered by Apache Spark, MLlib, and Streamlit")
```

```python
import streamlit as st
import pandas as pd
import plotly.express as px
import networkx as nx
import matplotlib.pyplot as plt
import os

# Title
st.title("Twitter Political Discourse Analysis")

# Load sentiment data from mounted Google Drive path
@st.cache_data
def load_data():
    path = "/content/drive/MyDrive/twitter_data/processed_sentiment.csv"
    if not os.path.exists(path):
        st.error("Sentiment data not found in Drive at: twitter_data/processed_sentiment.csv")
        return pd.DataFrame(columns=["clean_text", "sentiment", "date"])
    df = pd.read_csv(path)
    df["date"] = pd.to_datetime(df["date"], errors='coerce')
    df = df.dropna(subset=["date", "sentiment"])
    return df

df = load_data()

if df.empty:
    st.stop()

# Sidebar filters
st.sidebar.header("Filters")
selected_sentiment = st.sidebar.multiselect(
    "Select sentiment",
    options=df["sentiment"].dropna().unique().tolist(),
    default=df["sentiment"].dropna().unique().tolist()
)

# Filter data
filtered_df = df[df["sentiment"].isin(selected_sentiment)].copy()

# Ensure 'date' column is datetime and no NaT rows are present
filtered_df["date"] = pd.to_datetime(filtered_df["date"], errors="coerce")
filtered_df = filtered_df.dropna(subset=["date"])

# Sentiment by hour
st.subheader("Sentiment Trend by Hour")
filtered_df["hour"] = filtered_df["date"].dt.hour
hour_fig = px.histogram(
    filtered_df,
    x="hour",
    color="sentiment",
    barmode="group",
    title="Sentiment by Hour of Day"
)
st.plotly_chart(hour_fig)

# Sentiment distribution
st.subheader("Overall Sentiment Distribution")
dist_fig = px.histogram(
    filtered_df,
    x="sentiment",
    color="sentiment",
    title="Sentiment Distribution"
)
st.plotly_chart(dist_fig)

# Load influencer edges from Drive
@st.cache_data
def load_influencers():
    path = "/content/drive/MyDrive/twitter_data/user_edges.csv"
    if not os.path.exists(path):
        return pd.DataFrame(columns=["source", "target"])
    return pd.read_csv(path)

gdf = load_influencers()

st.subheader("Top Influencers")
if not gdf.empty:
    G = nx.DiGraph()
    G.add_edges_from(zip(gdf["source"], gdf["target"]))
    centrality = nx.degree_centrality(G)
    top_influencers = sorted(centrality.items(), key=lambda x: x[1], reverse=True)[:5]

    for user, score in top_influencers:
        st.write(f"{user}: {score:.4f}")

    st.subheader("Influencer Network")
    fig, ax = plt.subplots()
    pos = nx.spring_layout(G)
    nx.draw(G, pos, with_labels=True, node_color='skyblue', edge_color='gray',
            node_size=1500, font_size=10, ax=ax)
    st.pyplot(fig)
else:
    st.info("No influencer data available. Make sure 'user_edges.csv' is in Drive under 'twitter_data'.")

st.caption("Powered by Apache Spark, MLlib, and Streamlit")
```
